```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import polars as pl

cols = ['Experiment', 'study', 'condition_id', 'rep_id', 'creator', 'run_date', 'Strain Description', 
 'Base Media', 'Carbon Source (g/L)', 'Nitrogen Source (g/L)', 'Electron Acceptor', 'Trace Element Mixture', 
 'Supplement', 'Temperature (C)', 'pH', 'Growth Rate (1/hr)']

sns.set(style="ticks", palette="Set2")

pl.Config(tbl_rows=30)
```

```python
m = pl.read_csv("external_data/PRECISE2_metadata.tsv", separator="\t")
d = pl.read_csv("external_data/PRECISE2_counts.csv")
d
```

```python
# melt data
dm = d.melt(id_vars="Geneid", variable_name="Experiment", value_name="Gene_count")
dj = dm.join(m, on="Experiment", how="left", coalesce=True)

cols = ['Experiment', 'Geneid', 'Gene_count', 'study', 'condition_id', 'rep_id', 'creator', 'run_date', 'Strain Description', 
 'Base Media', 'Carbon Source (g/L)', 'Nitrogen Source (g/L)', 'Electron Acceptor', 'Trace Element Mixture', 
 'Supplement', 'Temperature (C)', 'pH', 'Growth Rate (1/hr)']

dj.select(cols)
```

```python
plt.figure(figsize=(6, 3))

sns.stripplot(y='Geneid', x="Gene_count", 
              data = dj.filter((pl.col('Geneid').is_in(["b3988", "b3942"])) & (pl.col("Gene_count") > 0))
              .with_columns(pl.col("Geneid").replace({"b3988": "rpoC", "b3942": "katG"}))
              .to_pandas(), 
            log_scale=True,  
            size=3,
            # common_norm=False, legend=True, 
            # kind="hist"
              )

# Set the x-axis label
plt.ylabel('Gene ID')
plt.xlabel('Transcript per million (TPM)')

# set x axis limit to 10^3 to 10^6
plt.xlim(10**2, 10**6)

# save as svg
plt.savefig("figures/katG_rpoC_expression.svg", dpi=300, bbox_inches='tight')

# save as png
plt.savefig("figures/katG_rpoC_expression.png", dpi=300, bbox_inches='tight')
# 
```
